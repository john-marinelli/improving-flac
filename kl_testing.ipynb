```python
import numpy as np
```

```python
rng = np.random.default_rng(seed=0)

SR = 44100
duration = 2.0 
sigma = 0.1
mu = 0.0

length = int(SR * duration)
noise = rng.normal(loc=mu, scale=sigma, size=length)
```

```python
n = 64
N = -(-len(noise) // n)
X = np.pad(noise, (0, N * n - len(noise))).reshape(N, n)
```

```python
R = np.cov(X, rowvar=False)
```

```python
lam, U = np.linalg.eigh(R)
```

```python
Y = X @ U
```

```python
assert np.allclose(np.cov(Y, rowvar=False), np.diag(lam))
```

```python
mae = np.mean(np.abs(Y))
```
